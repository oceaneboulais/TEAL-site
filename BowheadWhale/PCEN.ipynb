```python
import librosa
import librosa.display
import numpy as np
import torch
import torchaudio
import matplotlib.pyplot as plt

# Example: Load a sample audio file (replace file path with your data)
y, sr = librosa.load("your_audio.wav", sr=None)

# Generate a log-mel spectrogram
S = librosa.feature.melspectrogram(y, sr=sr, n_mels=64, fmax=500)  # restrict fmax as needed
S_dB = librosa.power_to_db(S, ref=np.max)

plt.figure(figsize=(10, 4))
librosa.display.specshow(S_dB, sr=sr, x_axis='time', y_axis='mel', fmax=500)
plt.colorbar(format='%+2.0f dB')
plt.title('Log-Mel Spectrogram')
plt.tight_layout()
plt.show()
```
